```python
import pandas as pd
from datetime import datetime
```

```python
# Q1
dt = datetime(2026, 2, 15, 10, 30)
print(dt.year)
print(dt.month)
print(dt.day)
print(dt.hour)
print(dt.minute)
```

```python
# Q2
start = datetime(2026, 2, 15, 10, 0)
end = datetime(2026, 2, 15, 12, 30)
difference = end - start
print(difference)
print(difference.total_seconds() // 3600, "hours")
print((difference.total_seconds() % 3600) // 60, "minutes")
```

```python
# Q3
date_string = "20/02/2026"
date_obj = datetime.strptime(date_string, "%d/%m/%Y")
print(date_obj.strftime("%Y-%m-%d"))
```

```python
# Q4
dates = ["2026-02-01", "2026-02-05", "2026-02-10"]
print(pd.to_datetime(dates))
```

```python
# Q5
dates = ["2026-02-01", "invalid", "2026-02-05", None]
date_values = pd.to_datetime(dates, errors="coerce")
print(date_values)
print(pd.isna(date_values))
```

```python
# Q6
dates = pd.to_datetime(["2026-02-01", "2026-02-02", "2026-02-04",
                        "2026-02-06", "2026-02-08"])
sales = pd.Series([100, 150, 120, 180, 200], index=dates, name="units_sold")
print(sales.index)
print(type(sales.index))
```

```python
# Q7
print(sales.loc["2026-02-04"])
```

```python
# Q8
print(sales.iloc[0])
print(sales.iloc[-1])
```

```python
# Q9
print(sales.loc["2026-02-02":"2026-02-06"])
```

```python
# Q10
print("Total:", sales.sum())
print("Average:", sales.mean())
```

```python
# Q11
print(pd.date_range(start="2026-03-01", periods=10, freq="D"))
```

```python
# Q12
print(pd.date_range(end="2026-03-15", periods=5, freq="D"))
```

```python
# Q13
print(pd.date_range(start="2026-02-01", periods=6, freq="2D"))
```

```python
# Q14
print(pd.date_range(start="2026-02-01 09:00",
                    end="2026-02-01 15:00", freq="h"))
```

```python
# Q15
print(pd.date_range(start="2026-01-01", periods=6, freq="ME"))
```

```python
# Q16
print(pd.date_range(start="2026-02-06", periods=5, freq="B"))
```

```python
# Q17
dates = pd.date_range("2026-02-01", periods=6, freq="D")
daily_sales = pd.Series([100, 120, 150, 130, 180, 200], index=dates)
df = pd.DataFrame({"Current Sales": daily_sales,
                   "Previous Day Sales": daily_sales.shift(1)})
print(df)
```

```python
# Q18
df = pd.DataFrame({"Current Sales": daily_sales,
                   "Next Day Sales": daily_sales.shift(-1)})
print(df)
```

```python
# Q19
daily_change = daily_sales - daily_sales.shift(1)
print(daily_change)
```

```python
# Q20
shifted_sales = daily_sales.shift(2, freq="D")
print("Original Series:")
print(daily_sales)
print("Shifted Series:")
print(shifted_sales)
```
